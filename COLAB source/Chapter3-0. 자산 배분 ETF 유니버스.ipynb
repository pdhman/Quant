```python
import pandas as pd
import yfinance as yf
from datetime import datetime, timedelta
import os

from google.colab import drive
drive.mount('/content/drive')
```

```python
file_path = '/content/ETF 유니버스.csv'

# --- 1. 유니버스 정의 ---
universe = [
    # US Market
    'SPY', 'QQQ', 'IWM', 'DIA', 'VUG', 'IWO', 'VTV', 'IWN', 'USMV', 'QUAL', 'MTUM', 'RSP','VTI', 'VBR',
    # US Sectors
    'XLK', 'XLV', 'XLF', 'XLY', 'XLP', 'XLE', 'XLI', 'XLB', 'XLU', 'XLRE', 'XLC',
    'SMH', 'IBB', 'KBE', 'XOP', 'ITB',
    # Global
    'EFA', 'EEM', 'ILF', 'AAXJ', 'VGK', 'EWG', 'EWU', 'EWQ', 'VNM','VEA', 'VWO',
    'EWJ', 'MCHI', 'INDA', 'EWY', 'EWT', 'VNM', 'EWZ', 'EWW', 'EWC', 'TUR',
    # Fixed Income
    'SHV', 'SHY', 'IEF', 'TLT', 'GOVT', 'ZROZ', 'TIP', 'VTIP',
    'LQD', 'HYG', 'SJNK', 'BNDX', 'EMB', 'CWB','IEI','BND','AGG', 'UST',
    # Commodities
    'DBC', 'PDBC', 'GLD', 'IAU', 'SLV', 'PPLT','GSG',
    'USO', 'UNG', 'URA', 'CPER', 'DBA', 'LIT',
    # Alternatives & Thematic
    'VNQ', 'VNQI', 'REM', 'DBMF', 'KMLM', 'WTMF', 'PSP', 'UUP', 'FXE',
    'IGV', 'CIBR', 'BOTZ', 'TAN', 'PAVE', 'ARKK',
    # Crypto
    'BITO', 'IBIT', 'ETHE', 'WGMI', 'BLOK',
    # Cash
    'BIL', 'SGOV', 'NEAR', 'MINT', 'ICSH', 'JPST', 'PULS',
    # Ritz
    'VNQ'
]

# --- 2. 데이터 업데이트 함수  ---
def update_asset_data(file_path, target_tickers):
    # 초기 시작일 (파일이 아예 없을 때, 혹은 신규 종목의 과거 데이터를 가져올 때 사용)
    DEFAULT_START_DATE = '2005-01-01'

    # 1. 기존 데이터 불러오기
    if os.path.exists(file_path):
        try:
            df_old = pd.read_csv(file_path, index_col=0, parse_dates=True)
            print(f"기존 데이터 로드 완료: {len(df_old)}행, {len(df_old.columns)}종목")
            last_date = df_old.index[-1]
            existing_tickers = df_old.columns.tolist()
        except Exception as e:
            print(f"파일 읽기 오류: {e}. 새로 다운로드합니다.")
            df_old = pd.DataFrame()
            last_date = None
            existing_tickers = []
    else:
        print("기존 파일이 없습니다. 전체 데이터를 새로 다운로드합니다.")
        df_old = pd.DataFrame()
        last_date = None
        existing_tickers = []

    # 2. 신규 티커 확인 (타겟 리스트에는 있는데 파일에는 없는 종목)
    # set 연산을 통해 차집합 구하기
    new_tickers = list(set(target_tickers) - set(existing_tickers))

    # 3. [Step A] 신규 티커의 '과거 데이터' 백필 (Back-fill)
    # 기존 파일이 있고, 마지막 날짜가 존재하며, 새로운 티커가 있을 경우에만 실행
    if last_date is not None and new_tickers:
        print(f"🆕 새로운 종목 {len(new_tickers)}개 발견! ({new_tickers[:3]}...) 과거 데이터를 병합합니다.")
        try:
            # 신규 종목만 과거부터 last_date까지 다운로드
            df_new_history = yf.download(new_tickers, start=DEFAULT_START_DATE, end=last_date, auto_adjust=True, progress=False)

            # 멀티인덱스 처리 ('Close'만 추출)
            if isinstance(df_new_history.columns, pd.MultiIndex):
                try:
                    df_new_history = df_new_history['Close']
                except KeyError:
                     # 'Close'가 없으면 Adj Close 시도
                    if 'Adj Close' in df_new_history.columns.get_level_values(0):
                        df_new_history = df_new_history['Adj Close']

            # 기존 데이터(df_old)에 옆으로 붙이기 (Merge/Concat axis=1)
            # 인덱스(날짜)를 기준으로 합칩니다.
            df_old = pd.concat([df_old, df_new_history], axis=1)
            print("신규 종목 과거 데이터 병합 완료.")

        except Exception as e:
            print(f"신규 종목 과거 데이터 다운로드 실패: {e}")

    # 4. [Step B] 전체 티커의 '최신 데이터' 업데이트 (Gap Filling)
    # 마지막 날짜가 있으면 그 다음날부터, 없으면 초기 날짜부터
    if last_date:
        start_date = last_date + timedelta(days=1)
    else:
        start_date = datetime.strptime(DEFAULT_START_DATE, "%Y-%m-%d")

    end_date = datetime.now()

    if start_date > end_date:
        print("이미 최신 데이터입니다.")
        # 신규 티커만 추가되었을 수도 있으므로 저장 후 리턴
        if new_tickers:
            df_old.to_csv(file_path, float_format='%.2f')
            print("파일이 업데이트(신규 종목 추가) 되었습니다.")
        return df_old

    print(f"🔄 {start_date.date()} ~ {end_date.date()} 기간 데이터를 업데이트합니다...")

    try:
        # 전체 타겟 티커에 대해 최신 데이터 다운로드
        df_recent = yf.download(target_tickers, start=start_date, end=end_date, auto_adjust=True, progress=False)

        if not df_recent.empty:
            # 멀티인덱스 처리
            if isinstance(df_recent.columns, pd.MultiIndex):
                try:
                    df_recent = df_recent['Close']
                except KeyError:
                    if 'Adj Close' in df_recent.columns.get_level_values(0):
                        df_recent = df_recent['Adj Close']

            # 5. 기존 데이터와 최신 데이터 세로로 합치기 (Concat axis=0)
            df_updated = pd.concat([df_old, df_recent], axis=0)
        else:
            print("받아올 최신 데이터가 없습니다 (휴장일 등).")
            df_updated = df_old

    except Exception as e:
        print(f"데이터 다운로드 중 오류 발생: {e}")
        return df_old

    # 6. 데이터 정리 (중복 제거 및 정렬)
    # 인덱스 중복 제거 (keep='last'로 최신 값 유지)
    df_updated = df_updated[~df_updated.index.duplicated(keep='last')]
    # 날짜순 정렬
    df_updated.sort_index(inplace=True)
    # 유니버스에 있는 컬럼만 남기고 순서 정렬 (필터링)
    # 데이터에는 있지만 유니버스에서 뺀 종목이 있다면 여기서 제거됨
    final_cols = [t for t in target_tickers if t in df_updated.columns]
    df_updated = df_updated[final_cols]

    # 7. 결과 저장
    df_updated.to_csv(file_path, float_format='%.2f')

    print("-" * 30)
    print(f"업데이트 완료! 총 {len(df_updated)}행, {len(df_updated.columns)}종목")
    print("-" * 30)

    return df_updated

# --- 실행 블록 ---
if __name__ == "__main__":
    file_path = '/content/drive/MyDrive/Colab Notebooks/ETF 유니버스.csv'

    # 100개 유니버스를 인자로 전달
    updated_df = update_asset_data(file_path, universe)

    if updated_df is not None:
        print("\n[데이터 미리보기]")
        print(updated_df.iloc[-5:, :5]) # 마지막 5일, 앞 5개 종목만 출력
```
